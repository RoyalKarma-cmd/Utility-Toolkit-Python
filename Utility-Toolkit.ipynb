```python
import random
from datetime import datetime
```

```python
while True:
     print("\n----Utility Toolkit----")
     print("1. Calculator")
     print("2. Random Number Generator")
     print("3. Dice Roller")
     print("4. Current Date & Time")
     print("5. Age calculator")
     print("6. Exit")

     Choice = int(input("Enter the choice: "))

     if Choice == 1:
          
          num1 = int(input("Enter the first number: "))
          num2 = int(input("ENter the second number: "))

          operations = input("Enter the operaions(+, -, *, /, %, ^2, percentage, power, ^3): ")

          Result = {num1, num2}
          if operations == "+":
              print("The ans of num1 & num2 is: " , num1 + num2)

          elif operations == "-":
              print("The ans of num1 & num2 is: " ,num1 - num2)

          elif operations == "*":
              print("The ans of num1 & num2 is: " , num1 * num2)

          elif operations == "/":
              if num2 == 0:
                  print("Cant't Divide by 0")
              else:
                  print("The ans of num1 & num2 is: " , num1 / num2)

          elif operations == "%":
              print("The ans of num1 & num2 is: " , num1 % num2)

          elif operations == "^2":
              print("The ans of num1 is: ", num1 ** 2)
              print("The ans of num2 is: ", num2 ** 2)

          elif operations == "percentage":
              print("The ans of num1 & num2 is: ", (num1/num2) * 100)

          elif operations == "power":
              print("The ans of num1 & num2 is: ", num1 ** num2)

          elif operations == "^3":
              print("The ans of num1 is: ", num1 ** 3)
              print("The ans of num2 is: ", num2 ** 3)

          else:
              print("invalid operation/input")

     
     elif Choice == 2:
         
         Maximum = int(input("Enter the Minimum number: "))
         Minimum = int(input("ENter the Maximum number: "))

         print("Generated number is: ", random.randint(Maximum, Minimum))

     elif Choice == 3:
         print("Roll the Dice: ", random.randint(1, 6))

     elif Choice == 4:
          
          now = datetime.now()
          
          print("Current Date is: ", now.strftime("%d - %m - %Y"))
          
          print("Current time is: ", now.strftime("%H:%M:%S"))

          print("Current day is: ", now.strftime("%A"))

          print("Current Month is: ", now.strftime("%B"))

     elif Choice == 5:
         
         Birth_Year = int(input("Enter the Birth Year: "))
         Current_Year = datetime.now().year
         Age = Current_Year - Birth_Year  

         print("Your Age is: ",Age)

     elif Choice == 6:
         print("Thanks for Using Utility Tool Box")
         break

     else:
         
         print("Invalid Input")
                  
     

```
